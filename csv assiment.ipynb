```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_theme(style="whitegrid")
```

```python
loan_df = pd.read_csv('Downloads/loan.csv')

loan_df.head()
```

```python

print("--- Dataset ki details ---")
loan_df.info()

print("\n--- Khaali (Missing) Values ---")
print(loan_df.isnull().sum())
```

```python
text_cols = loan_df.select_dtypes(include=['object']).columns
for col in text_cols:
    if loan_df[col].isnull().sum() > 0:
        loan_df[col] = loan_df[col].fillna(loan_df[col].mode()[0])

num_cols = loan_df.select_dtypes(include=['number']).columns
for col in num_cols:
    if loan_df[col].isnull().sum() > 0:
        loan_df[col] = loan_df[col].fillna(loan_df[col].median())

print("--- Cleaning ke baad bachi hui Missing Values ---")
print(loan_df.isnull().sum())
```

```python
# Udaharan ke liye 'Loan_Status' ka graph (Aap column ka naam badal sakte hain)
plt.figure(figsize=(6,4))
sns.countplot(data=loan_df, x='Loan_Status', palette='Set1')
plt.title('Loan Status ki Tafseel')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

cat_cols = loan_df.select_dtypes(include=['object']).columns
for col in cat_cols:
    plt.figure(figsize=(6, 4))
    sns.countplot(data=loan_df, x=col, palette='Set2')
    plt.title(col + ' ki Distribution')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

num_cols = loan_df.select_dtypes(include=['number']).columns
for col in num_cols:
    plt.figure(figsize=(7, 4))
    sns.histplot(loan_df[col], kde=True, color='purple', bins=30)
    plt.title(col + ' ki Distribution (Histogram)')
    plt.tight_layout()
    plt.show()
```

```python
adult_df = pd.read_csv('Downloads/adult.csv')

adult_df.head()
```

```python
print("--- Adult Dataset Info ---")
adult_df.info()

print("\n--- Missing Values ---")
print(adult_df.isnull().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))
sns.histplot(adult_df['age'], bins=30, kde=True, color='teal')
plt.title('Adult Dataset: Age Column ki Distribution', fontsize=14)
plt.xlabel('Age (Umr)', fontsize=12)
plt.ylabel('Count (Tadaad)', fontsize=12)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
sns.boxplot(x=adult_df['age'], color='lightgreen')
plt.title('Adult Dataset: Age Column ka Boxplot', fontsize=14)
plt.xlabel('Age (Umr)', fontsize=12)
plt.tight_layout()
plt.show()

print("--- Age Column ki Statistical Summary ---")
print(adult_df['age'].describe())
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

adult_df.replace('?', np.nan, inplace=True)

cat_cols_adult = adult_df.select_dtypes(include=['object']).columns
for col in cat_cols_adult:
    if adult_df[col].isnull().sum() > 0:
        adult_df[col] = adult_df[col].fillna(adult_df[col].mode())

num_cols_adult = adult_df.select_dtypes(include=['number']).columns
for col in num_cols_adult:
    if adult_df[col].isnull().sum() > 0:
        adult_df[col] = adult_df[col].fillna(adult_df[col].median())

print("--- Adult Dataset Cleaning Done Without Error! ---")
print("Bachi hui missing values:", adult_df.isnull().sum().sum())


for col in cat_cols_adult:
    plt.figure(figsize=(8, 4))
    sns.countplot(data=adult_df, x=col, order=adult_df[col].value_counts().index[:7], palette='muted')
    plt.title(f'Adult Dataset: {col} ki Distribution')
    plt.xticks(rotation=30, ha='right')
    plt.tight_layout()
    plt.show()


for col in num_cols_adult:
    if col != 'age': 
        plt.figure(figsize=(8, 4))
        sns.histplot(adult_df[col], kde=True, bins=30, color='orange')
        plt.title(f'Adult Dataset: {col} ka Histogram')
        plt.tight_layout()
        plt.show()
```

```python
hotel_df = pd.read_csv('Downloads/hotel_booking.csv')

cat_cols_hotel = hotel_df.select_dtypes(include=['object']).columns
for col in cat_cols_hotel:
    if hotel_df[col].isnull().sum() > 0:
        hotel_df[col] = hotel_df[col].fillna(hotel_df[col].mode())

num_cols_hotel = hotel_df.select_dtypes(include=['number']).columns
for col in num_cols_hotel:
    if hotel_df[col].isnull().sum() > 0:
        hotel_df[col] = hotel_df[col].fillna(hotel_df[col].median())

print("--- Hotel Booking Dataset Cleaning Complete! ---")
print("Bachi hui missing values:", hotel_df.isnull().sum().sum())
```

```python
for col in cat_cols_hotel:
    plt.figure(figsize=(8, 4))
    sns.countplot(data=hotel_df, x=col, order=hotel_df[col].value_counts().index[:5], palette='pastel')
    plt.title(f'Hotel Dataset: {col} ki Distribution')
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()
sample_num_cols = [c for c in num_cols_hotel if c in ['lead_time', 'stays_in_weekend_nights', 'stays_in_week_nights', 'adults']][:4]

for col in sample_num_cols:
    plt.figure(figsize=(8, 4))
    sns.histplot(hotel_df[col], kde=True, bins=30, color='skyblue')
    plt.title(f'Hotel Dataset: {col} ka Histogram')
    plt.tight_layout()
    plt.show()
```
